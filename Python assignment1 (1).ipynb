3. String Methods:

```python
strM = "Python beginner tutorial"
```

Convert the sentence to uppercase.

```python
print(strM.upper())
```

Convert the sentence to lowercase.

```python
print(strM.lower())
```

Use Capitalize and return the sentence to the original input form.

```python
print(strM.capitalize())
```

Count the total number of occurrences of character ‘t’ in the string.

```python
print(strM.count('t'))
```

Replace all occurrences of “Python” with “Machine Learning” in the input string
strM = “Python beginner tutorial”

```python
print(strM.replace("Python", "Machine Learning"))
```

Tuples (Creation, Modification and Access) :

Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)
```

Concatenate the two tuples and store it in “t_combine”

```python
t_combine=t1+t2
print(t_combine)
```

Repeat the elements of “t_combine” 3 times

```python
print(t_combine*3)
```

Access the 3rd element from “t_combine”

```python
print(t_combine[2])
```

Access the first three elements from “t_combine”

```python
print(t_combine[0:3])
```

Access the last three elements from “t_combine”

```python
print(t_combine[-3:])
```
